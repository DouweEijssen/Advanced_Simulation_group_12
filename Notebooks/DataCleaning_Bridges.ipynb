# Conversion Panda dataframe
The roads.tsv is in wide formate. Since we need the information about the chainage of the road LRPs, we should update the document Roads_InfoAboutEachLRP with the processed _roads.tsv (see DataCleaning_Roads notebook). Such that the updates latitudes and longitudes of the LRP's are in the document Road_InfoAboutEachLRP.

This is the approach we aim to do:

1. _roads.tsv is transferred to a long format
2. Road_InfoAboutEachLRP to panda dataframe
2. Merge the column 'lat' and 'lon' from the _roads.tsv into the Road_InfoAboutEachLRP (this step is not executed)

```python
cols = roads_overview.columns.tolist()

## rename unnamed columns
start = cols.index("lon2") + 1
tail = cols[start:]

new_tails = []
idx = 3
for i in range(0, len(tail), 3):
    new_tails.extend([f"lrp{idx}", f"lat{idx}", f"lon{idx}"])
    idx += 1

new_cols = cols[:start] + new_tails
new_cols

roads = roads_overview.copy()
roads.columns = new_cols

# Reshape wide -> long
long = pd.wide_to_long(
    roads,
    stubnames=["lrp", "lat", "lon"],
    i="road",
    j="idx",
    sep="",
    suffix=r"\d+"
).reset_index()

# Clean rows
roads_long = long[long["lrp"].notna() & (long["lrp"].astype(str).str.strip() != "")]
roads_long["lat"] = pd.to_numeric(long["lat"], errors="coerce")
roads_long["lon"] = pd.to_numeric(long["lon"], errors="coerce")
```

# algorithms checking distances
Still some fine tuning needed, because it doesn't work well for every road. Maybe we need to work with a percentage treshold or something.

```python
def point_to_segment_distance(lat, lon, lat1, lon1, lat2, lon2):
    """
    Calculate perpendicular distance from a point to a line segment (in meters)
    using geodesic distances for better accuracy.
    """
    # Convert points to 2D Cartesian approx for short distances
    # Using equirectangular projection
    R = 6371000  # Earth radius in meters
    x1 = np.radians(lon1) * R * np.cos(np.radians((lat1 + lat2) / 2))
    y1 = np.radians(lat1) * R
    x2 = np.radians(lon2) * R * np.cos(np.radians((lat1 + lat2) / 2))
    y2 = np.radians(lat2) * R
    x0 = np.radians(lon) * R * np.cos(np.radians((lat1 + lat2) / 2))
    y0 = np.radians(lat) * R

    dx = x2 - x1
    dy = y2 - y1
    if dx == 0 and dy == 0:
        # segment is a point
        return np.hypot(x0 - x1, y0 - y1)

    t = np.clip(((x0 - x1) * dx + (y0 - y1) * dy) / (dx**2 + dy**2), 0, 1)
    proj_x = x1 + t * dx
    proj_y = y1 + t * dy
    return np.hypot(x0 - proj_x, y0 - proj_y)
```

```python
def match_bridges_to_lrps(
    bridges_df,
    roads_df,
    road_name,
    chainage_col='chainage',
    lat_col='lat',
    lon_col='lon',
    lrp_col='lrp',
    road_col='road',
    threshold_m=25
):
    """
    Match bridges to bounding LRPs and flag out-of-range bridges using perpendicular distance.
    """
    # --- Filter & sort once
    bridges = (
        bridges_df[bridges_df[road_col] == road_name]
        .sort_values(chainage_col)
        .reset_index(drop=False)
    )

    roads = (
        roads_df[roads_df[road_col] == road_name]
        .sort_values(chainage_col)
        .reset_index(drop=False)
    )

    road_chainage = roads[chainage_col].to_numpy()
    results = []

    for bridge_idx, bridge in bridges.iterrows():
        bridge_chainage = bridge[chainage_col]

        idx = np.searchsorted(road_chainage, bridge_chainage, side='right')

        # --- Out-of-bounds
        if idx == 0 or idx >= len(road_chainage):
            results.append({
                road_col: road_name,
                'bridge_index': bridge_idx,
                'bridge_lrp': bridge.get('LRPName'),
                'status': 'out-of-range',
                lon_col: bridge[lon_col],
                lat_col: bridge[lat_col]
            })
            continue

        start, end = idx - 1, idx

        # Coordinates
        s_lat, s_lon = roads.loc[start, [lat_col, lon_col]]
        e_lat, e_lon = roads.loc[end, [lat_col, lon_col]]
        b_lat, b_lon = bridge[lat_col], bridge[lon_col]

        # --- Perpendicular distance to segment
        perp_dist = point_to_segment_distance(b_lat, b_lon, s_lat, s_lon, e_lat, e_lon)

        status = 'out-of-range' if perp_dist > threshold_m else 'ok'

        results.append({
            road_col: road_name,
            'start_LRPName': roads.loc[start, lrp_col],
            'end_LRPName': roads.loc[end, lrp_col],
            'start_chainage': roads.loc[start, chainage_col],
            'end_chainage': roads.loc[end, chainage_col],
            'bridge_index': bridge_idx,
            'bridge_lrp': bridge.get('LRPName'),
            'perp_distance_m': perp_dist,
            'status': status,
            lon_col: b_lon,
            lat_col: b_lat
        })

    return pd.DataFrame(results)
```

```python
~!import matplotlib.pyplot as plt
import pandas as pd


def plot_bridge_status(
    roads_df,
    results_df,
    road_name=None,
    road_col='road',
    status_col='status',
    ok_label='ok',
    bad_label='out-of-range',
    lon_col='lon',
    lat_col='lat',
    figsize=(9, 7),
    show=True
):
    """
    Plot road geometry with bridges colored by status.

    Parameters
    ----------
    roads_df : pandas.DataFrame
        Road geometry (lat/lon)
    results_df : pandas.DataFrame
        Bridge matching results with status
    road_name : str or None
        Road to plot (None = plot all roads)
    """

    # --- Filter by road (optional)
    if road_name is not None:
        road = roads_df[roads_df[road_col] == road_name]
        bridges = results_df[results_df[road_col] == road_name]
        title = f"Bridge status for road {road_name}"
    else:
        road = roads_df
        bridges = results_df
        title = "Bridge status (all roads)"

    if road.empty:
        raise ValueError("No road geometry found for the selected road(s).")

    # --- Split bridge status
    ok = bridges[bridges[status_col] == ok_label]
    bad = bridges[bridges[status_col] == bad_label]

    # --- Plot
    plt.figure(figsize=figsize)

    plt.plot(
        road[lon_col],
        road[lat_col],
        linewidth=2,
        label='Road'
    )

    if not ok.empty:
        plt.scatter(
            ok[lon_col],
            ok[lat_col],
            s=40,
            label=f'OK bridges ({len(ok)})'
        )

    if not bad.empty:
        plt.scatter(
            bad[lon_col],
            bad[lat_col],
            s=40,
            label=f'Out-of-range bridges ({len(bad)})'
        )

    plt.title(title)
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    if show:
        plt.show()
```
