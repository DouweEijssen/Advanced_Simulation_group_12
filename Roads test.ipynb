Calculate distances:

* make algorithm that for each road calculates the distances

```python
import numpy as np

def get_lrp_distances(df):
    """
    Returns a pandas Series with haversine distance (km)
    between each LRP and the next one.
    Does NOT modify the dataframe.
    """
    R = 6371.0  # Earth radius in km

    lat1 = np.radians(df['LAT'])
    lon1 = np.radians(df['LON'])
    lat2 = np.radians(df['LAT'].shift(-1))
    lon2 = np.radians(df['LON'].shift(-1))

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

    return R * c
```

```python
# --- ensure clean neighbour indexing ---
df_rd = df_rd.reset_index(drop=True)

# --- plot BEFORE ---
plot_rd(rd_name + " (before)", df_rd)

# --- compute distances ---
lrp_dist = get_lrp_distances(df_rd)
threshold = lrp_dist.quantile(0.75)*20

# --- find bad edges ---
bad_edges = lrp_dist > threshold

# --- the bad LRP is usually the NEXT point ---
bad_idx = [i+1 for i in bad_edges[bad_edges].index]

# --- keep safe interior points only ---
bad_idx = [i for i in bad_idx if 0 < i < len(df_rd)-1]

# --- simple 3-point midpoint fix (Euler style) ---
for i in bad_idx:
    if i+1 in bad_idx:
        df_rd.loc[i, 'LAT'] = (df_rd.loc[i-1, 'LAT'] + df_rd.loc[i+1, 'LAT']) / 2
        df_rd.loc[i, 'LON'] = (df_rd.loc[i-1, 'LON'] + df_rd.loc[i+1, 'LON']) / 2
    elif i+2 in bad_idx:
        df_rd.loc[i, 'LAT'] = (df_rd.loc[i-1, 'LAT'] + df_rd.loc[i+2, 'LAT']) / 2
        df_rd.loc[i, 'LON'] = (df_rd.loc[i-1, 'LON'] + df_rd.loc[i+2, 'LON']) / 2
        df_rd.loc[i+1, 'LAT'] = (df_rd.loc[i-1, 'LAT'] + df_rd.loc[i+2, 'LAT']) / 2
        df_rd.loc[i+1, 'LON'] = (df_rd.loc[i-1, 'LON'] + df_rd.loc[i+2, 'LON']) / 2
    else:
        continue
# --- plot AFTER ---
plot_rd(rd_name + " (after)", df_rd)

print("Threshold:", threshold)
print("Bad count:", len(bad_idx))
```
