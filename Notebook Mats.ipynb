```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv('WBSIM_Lab1_2024/infrastructure/Roads_InfoAboutEachLRP.csv')
```

```python
unique_roads = df['road'].unique()
plt.figure(figsize=(10, 6))
for road_id in unique_roads[:3]:
    df_rd = df[df['road'] == road_id][['lat', 'lon', 'chainage']].reset_index(drop=True)

    #print road head
    print(road_id)
    print(df_rd.head())
    print(df_rd.shape)



    plt.figure()
    plt.plot(df_rd['lon'], df_rd['lat'], marker='o', ms=3)
    plt.title(road_id)
    plt.show()


```
